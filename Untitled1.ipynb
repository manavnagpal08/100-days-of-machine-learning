```python
#@title 1.4 — Download Pretrained Models
# Wav2Lip needs:
#   • wav2lip_gan.pth  — higher visual quality (recommended)
#   • wav2lip.pth      — faster inference, slightly lower quality
#   • s3fd.pth         — face detection backbone
#
# ⚠️  The original SharePoint links from the paper authors are dead (404).
# We pull from the community HuggingFace mirror: Nekochu/Wav2Lip
# using huggingface_hub which handles LFS, retries, and local caching.

import os, urllib.request, shutil
from pathlib import Path

# huggingface_hub is pre-installed on Colab; install only if missing
try:
    from huggingface_hub import hf_hub_download
except ImportError:
    os.system("pip install -q huggingface_hub")
    from huggingface_hub import hf_hub_download

CHECKPOINTS_DIR = Path("/content/Wav2Lip/checkpoints")
CHECKPOINTS_DIR.mkdir(exist_ok=True)

def download_hf(repo_id, filename, dest_path, label):
    """Download a model file from HuggingFace Hub with a direct-URL fallback."""
    if dest_path.exists() and dest_path.stat().st_size > 1_000_000:
        print(f"   ✅ {label} already present ({dest_path.stat().st_size // 1_048_576} MB), skipping.")
        return
    print(f"   ⬇️  Downloading {label} from {repo_id} (~436 MB, please wait)...")
    try:
        # hf_hub_download saves to a cache; local_dir copies it alongside
        hf_hub_download(
            repo_id=repo_id,
            filename=filename,
            repo_type="model",
            local_dir=str(dest_path.parent),
        )
        # The file lands at dest_path.parent/filename
        downloaded = dest_path.parent / filename
        if downloaded.resolve() != dest_path.resolve() and downloaded.exists():
            shutil.move(str(downloaded), str(dest_path))
        print(f"   ✅ {label} ready ({dest_path.stat().st_size // 1_048_576} MB)")
    except Exception as e:
        print(f"   ⚠️  huggingface_hub error: {e}")
        print(f"   🔄 Trying direct HTTPS fallback...")
        fallback = f"https://huggingface.co/{repo_id}/resolve/main/{filename}"
        urllib.request.urlretrieve(fallback, dest_path)
        print(f"   ✅ {label} downloaded via fallback URL")

# ── 1. Wav2Lip GAN model (best quality) ───────────────────────────────────────
download_hf(
    repo_id="Nekochu/Wav2Lip",
    filename="wav2lip_gan.pth",
    dest_path=CHECKPOINTS_DIR / "wav2lip_gan.pth",
    label="wav2lip_gan.pth",
)

# ── 2. Wav2Lip standard model (faster) ────────────────────────────────────────
download_hf(
    repo_id="Nekochu/Wav2Lip",
    filename="wav2lip.pth",
    dest_path=CHECKPOINTS_DIR / "wav2lip.pth",
    label="wav2lip.pth",
)

# ── 3. s3fd face-detection backbone ───────────────────────────────────────────
FACE_DET_DIR = Path("/content/Wav2Lip/face_detection/detection/sfd")
FACE_DET_DIR.mkdir(parents=True, exist_ok=True)
S3FD_DEST = FACE_DET_DIR / "s3fd.pth"

if not S3FD_DEST.exists():
    print("   ⬇️  Downloading s3fd face detector...")
    urllib.request.urlretrieve(
        "https://www.adrianbulat.com/downloads/python-fan/s3fd-619a316812.pth",
        S3FD_DEST,
    )
    print(f"   ✅ s3fd.pth saved ({S3FD_DEST.stat().st_size // 1_048_576} MB)")
else:
    print("   ✅ s3fd.pth already present, skipping.")

print("\n🎯 All models ready — proceed to Step 2.")
```

```python
#@title 3.1 — Generate TTS Audio with gTTS
#@markdown **Type your script below, then run this cell.**

SPEECH_TEXT = "Hello! I am your AI talking avatar. Wav2Lip is making my lips move in perfect sync with this speech. Isn't that amazing?" #@param {type:"string"}
LANGUAGE = "en" #@param ["en", "es", "fr", "de", "hi", "pt", "ja", "ko", "zh"]

import os
from pathlib import Path

# Install gTTS (lightweight, no WebSocket, works on Colab IPs)
os.system("pip install -q gtts")
from gtts import gTTS

AUDIO_PATH = Path("/content/inputs/speech.wav")
AUDIO_PATH.parent.mkdir(exist_ok=True)

print(f"🎙️  Synthesising speech (language: {LANGUAGE})")
print(f"📝 Text: {SPEECH_TEXT[:80]}{'...' if len(SPEECH_TEXT) > 80 else ''}")

# gTTS outputs MP3 natively → convert to 16kHz mono WAV for Wav2Lip
mp3_tmp = AUDIO_PATH.with_suffix(".mp3")
tts = gTTS(text=SPEECH_TEXT, lang=LANGUAGE, slow=False)
tts.save(str(mp3_tmp))

# Convert MP3 → WAV 16kHz mono (Wav2Lip requirement)
os.system(f"ffmpeg -y -i {mp3_tmp} -ar 16000 -ac 1 {AUDIO_PATH} -loglevel error")
mp3_tmp.unlink(missing_ok=True)

import subprocess
dur = subprocess.run(
    ["ffprobe","-v","error","-show_entries","format=duration",
     "-of","default=noprint_wrappers=1:nokey=1", str(AUDIO_PATH)],
    capture_output=True, text=True
).stdout.strip()

print(f"\n✅ Audio saved: {AUDIO_PATH}")
print(f"   Duration: {float(dur):.2f}s")
```

```python
#@title 4.1 — Configure Inference Settings

#@markdown **Model quality**
USE_GAN_MODEL = True #@param {type:"boolean"}
#@markdown *GAN = higher quality (recommended). Uncheck for faster but slightly blurrier output.*

#@markdown ---
#@markdown **Performance / quality trade-off**
RESIZE_FACTOR = "1" #@param ["1", "2", "4"]
#@markdown *Resize factor 1 = original resolution (slower). 2 or 4 = faster but lower res.*

SMOOTH_FACE = True #@param {type:"boolean"}
#@markdown *Smooth face crop boundaries to reduce jitter.*

#@markdown ---
#@markdown **Face crop padding (pixels, all sides)**
PAD_TOP    = 0  #@param {type:"integer"}
PAD_BOTTOM = 10 #@param {type:"integer"}
PAD_LEFT   = 0  #@param {type:"integer"}
PAD_RIGHT  = 0  #@param {type:"integer"}

# ── Resolve paths ──────────────────────────────────────────────────────────────
from pathlib import Path
import os

CHECKPOINT = (
    "/content/Wav2Lip/checkpoints/wav2lip_gan.pth"
    if USE_GAN_MODEL else
    "/content/Wav2Lip/checkpoints/wav2lip.pth"
)

AVATAR_VIDEO = "/content/inputs/avatar.mp4"
AUDIO_PATH   = "/content/inputs/speech.wav"
RESULTS_DIR  = Path("/content/results")
RESULTS_DIR.mkdir(exist_ok=True)
OUTPUT_VIDEO = RESULTS_DIR / "talking_avatar.mp4"

# Convert RESIZE_FACTOR from string → int for the inference command
RESIZE_FACTOR_INT = int(RESIZE_FACTOR)

for p, name in [(CHECKPOINT, "Model"), (AVATAR_VIDEO, "Avatar video"), (AUDIO_PATH, "Audio")]:
    if not Path(p).exists():
        raise FileNotFoundError(f"❌ {name} not found: {p}\nMake sure earlier steps completed successfully.")

print("⚙️  Inference configuration:")
print(f"   Model      : {'wav2lip_gan.pth (GAN)' if USE_GAN_MODEL else 'wav2lip.pth (standard)'}")
print(f"   Avatar     : {AVATAR_VIDEO}")
print(f"   Audio      : {AUDIO_PATH}")
print(f"   Output     : {OUTPUT_VIDEO}")
print(f"   Resize     : {RESIZE_FACTOR_INT}x")
print(f"   Smooth     : {SMOOTH_FACE}")
print(f"   Pad (T/B/L/R): {PAD_TOP}/{PAD_BOTTOM}/{PAD_LEFT}/{PAD_RIGHT}")
```

```python
#@title 4.2 — Run Wav2Lip (⏳ takes 1–5 min depending on video length)
import os, subprocess
from pathlib import Path

os.chdir("/content/Wav2Lip")

cmd = [
    "python", "inference.py",
    "--checkpoint_path", CHECKPOINT,
    "--face",           AVATAR_VIDEO,
    "--audio",          AUDIO_PATH,
    "--outfile",        str(OUTPUT_VIDEO),
    "--resize_factor",  str(RESIZE_FACTOR_INT),
    "--pads",           str(PAD_TOP), str(PAD_BOTTOM), str(PAD_LEFT), str(PAD_RIGHT),
]

print("🚀 Starting Wav2Lip inference...")
print("─" * 60)

# Capture both stdout and stderr so we can show the real error
result = subprocess.run(cmd, capture_output=True, text=True)

print(result.stdout)   # Always print stdout
if result.stderr:
    print("STDERR:\n", result.stderr)  # Print stderr so we see the actual crash

print("─" * 60)

if result.returncode != 0:
    print("❌ Inference failed — real error shown above ↑")
else:
    size_mb = Path(OUTPUT_VIDEO).stat().st_size / 1024 / 1024
    print(f"✅ Done! Output: {OUTPUT_VIDEO} ({size_mb:.2f} MB)")
```
